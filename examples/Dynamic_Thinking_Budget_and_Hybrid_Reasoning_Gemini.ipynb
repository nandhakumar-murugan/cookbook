This recipe demonstrates how to control the depth of internal chain-of-thought reasoning using **Thinking Budget parameters** (`thinking_budget`) in Gemini reasoning models.

## Initialize Client & Select Model

```python
import os
try:
    from google.colab import userdata
    if "GEMINI_API_KEY" in userdata.get_keys():
        os.environ["GEMINI_API_KEY"] = userdata.get("GEMINI_API_KEY")
except (ImportError, AttributeError):
    pass

from google import genai
from google.genai import types

MODEL_ID = "gemini-2.0-flash-thinking-exp"  # @param {type:"string"}
client = genai.Client()
```

## 1. Configure Thinking Budget & Execute Reasoning

```python
challenge_prompt = (
    "In a distributed Byzantine Fault Tolerant (BFT) network of 3f + 1 nodes, "
    "prove mathematically how batch Merkle tree proofs maintain consistency under adversarial partitions."
)

# Set thinking_budget in ThinkingConfig
config = types.GenerateContentConfig(
    thinking_config=types.ThinkingConfig(
        thinking_budget=2048  # Correct thinking_budget parameter
    ),
    temperature=0.2,
)

response = client.models.generate_content(
    model=MODEL_ID,
    contents=challenge_prompt,
    config=config,
)

print("Reasoning Output:")
print(response.text[:600] + "...")
```
